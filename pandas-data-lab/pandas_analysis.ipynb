```python
import pathlib as pa

p = pa.Path("/content/drive/MyDrive/ev_market_2026.csv")
```

```python
import pandas as pd
cars = pd.read_csv(p)
```

```python
cars = cars.drop("Unnamed: 24",axis=1)
```

```python
cars["efficiency"] = cars["battery_capacity_kwh"]/cars["range_miles"]
cars["efficiency"].round(2)
```

```python
mean_price = cars.groupby("brand")["price_usd"].describe().plot(kind="bar",color="green",figsize=(10,5))
```

```python

# رسم متوسط السعر لكل ماركة
cars.groupby("brand")["price_usd"].mean().plot(kind='line', color='skyblue', figsize=(10,5))
```
